```python
import pandas as pd
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import root_mean_squared_error as rmse
```

```python
df = pd.read_csv("./datasets/Salary_dataset.csv")
```

```python
df.sample(n=5)
```

```python
X = df[['YearsExperience']]
```

```python
y = df['Salary']
```

```python
import matplotlib.pyplot as plt
plt.plot(X['YearsExperience'], y)
plt.grid(True, color = "grey", linewidth = "1.4", linestyle = "-.")

# Show the plot
plt.show()
```

```python
print("No of data points: ", df.shape[0])
print("Min and Max Value of YearsExperience: (", \
    df['YearsExperience'].min(), ", ", df['YearsExperience'].max(), ")" )
print("Min and Max Value of Salary: (", df['Salary'].min(), ", ", df['Salary'].max(), ")" )
```

```python
model = LinearRegression()
```

```python
model.fit(X,y)
```

```python
model.coef_
```

```python
model.intercept_
```

```python
y_hat = model.predict(X)
```

```python
y_hat
```

```python
type(y_hat)
```

```python
y_hat_int = [int(x) for x in y_hat]
```

```python
y_hat_int
```

```python
type(y_hat_int)
```

```python
df['y_hat'] = y_hat_int
```

```python
df.rename(columns={'YearsExperience': 'x', 'Salary': 'y'}, inplace=True)
```

```python
df.head()
```

```python
print("RMSE: %.2f" % rmse(y, y_hat))
```
